```python
import os
import pathlib

import cv2 as cv
import numpy as np
import sklearn
import tqdm

from video import Video
```

```python
STRIDE = 24
NUM_CLUSTERS = 4
RANDOM_STATE = 0
NAME = f"0418_color_{STRIDE}_{NUM_CLUSTERS}_{RANDOM_STATE}"
NAME
```

```python
base_dir = pathlib.Path(os.getcwd())
input_dir = base_dir / "input"
output_dir = base_dir / "output"
```

```python
video = Video(input_dir / "anime.mp4", verbose=True)
```

```python
out = np.empty((video.num_frames // STRIDE, NUM_CLUSTERS, 3), dtype=np.uint8)

for i in tqdm.trange(video.num_frames // STRIDE):
    frame = video[STRIDE*i]
    
    X = frame.reshape(-1, 3)
    kmeans = sklearn.cluster.KMeans(
        n_clusters=NUM_CLUSTERS,
        random_state=RANDOM_STATE,
    ).fit(X)
    
    centers = np.clip(kmeans.cluster_centers_.astype(np.uint8), 0, 255)
    out[i] = centers
```

```python
np.save(output_dir / NAME, out)
```
